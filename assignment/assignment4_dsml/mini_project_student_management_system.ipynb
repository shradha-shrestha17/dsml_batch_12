**Project Title:**
Student Management System Using OOP Concepts


**Problem Statement:**
Develop a Student Management System using Python that demonstrates
the concepts of Class, Object, Encapsulation, Inheritance, and Abstraction.
The system should allow users to store student information, calculate
grades, and display student details based on different student categories.

**Requirements:**
1. Create an abstract class Student
The abstract class should contain:


*   Student ID
*   Student Name
*   A private attribute __marks
*   A method to set marks
*   A method to get marks
*   An abstract method calculate_grade()
*   A method display_details()





```python
from abc import ABC, abstractmethod

class Student(ABC):
  def __init__(self, sid, name):
    self.sid = sid
    self.name = name
    self.__marks = 0

  def set_marks(self, marks):
    self.__marks = marks

  def get_marks(self):
    return self.__marks


  @abstractmethod
  def calculate_grade(self):
    pass


  def display_details(self):
    print(f"Student ID: {self.sid}")
    print(f"Student Name: {self.name}")
    print(f"Marks: {self.get_marks()}")
    print(f"Grade: {self.calculate_grade()}")


class Grade(Student):
  def calculate_grade(self):
      marks = self.get_marks()
      if marks>90 and marks<=100:
        return "A+"
      elif marks>80 and marks<=90:
        return "A"
      elif marks>70 and marks<=80:
        return "B+"
      elif marks>60 and marks<=70:
        return "B"
      elif marks>50 and marks<=60:
        return "C+"
      elif marks>40 and marks<=50:
        return "C"
      elif marks>33 and marks<=40:
        return "D"
      else:
        return "F"


student = Grade(101, "Sita")
student.set_marks(88)
student.display_details()

```
